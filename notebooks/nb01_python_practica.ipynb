## 3. Programación orientada a objetos (clases): 

La programación con clases cambia la forma de pensar: en lugar de centrarse solo en funciones, se centra en **objetos que agrupan datos y comportamientos.**

### El mismo promedio

La programación orientada a objetos reúne datos y comportamientos relacionados en objetos creados a partir de clases. 

En una clase podemos guardar los datos dentro del objeto. En la comparación observamos `__init__`, `self`, atributo y método.]


`````python
class Analizador:
    def __init__(self, datos):
        self.datos = datos

    def promedio(self):
        return sum(self.datos) / len(self.datos)

analisis = Analizador(datos) # Instancia
analisis.promedio()
`````

`````python
otro_analisis = Analizador([5, 15])
print(analisis.datos, analisis.promedio())
print(otro_analisis.datos, otro_analisis.promedio())
`````

#### 1. self: una referencia a la instancia actual de una clase.

- Permite acceder a los datos y métodos del propio objeto.
- Se utiliza dentro de una clase para consultar o modificar sus atributos.
- Por convención, aparece como primer parámetro de los métodos de instancia.
- Python proporciona automáticamente la instancia cuando llamamos al método desde un objeto.
    

`````python
class Identificador:
    def saludar(self):
        print("Hola")
`````

`````python
# Cuando creamos una instancia y ejecutamos:
persona = Identificador()
persona.saludar()
`````

`````python
# Esto es equivalente a:
Identificador.saludar(persona)
`````

#### 2. Atributos: propiedades del objeto

Los atributos son las propiedades o características que almacenan información de un objeto.

Representan los datos asociados a una instancia.
Pueden tener valores distintos para cada objeto.
Habitualmente se inicializan dentro del método __init__().
El método __init__() permite inicializar los atributos cuando creamos un objeto.

Por ejemplo:

`````python
class Identificador:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
`````

`````python
# Ahora podemos crear una instancia de la clase:
persona = Identificador("Luis", 29)
`````

Cuando ejecutamos esta instrucción, Python crea el objeto y llama a `__init__()` para inicializar sus atributos.

De forma simplificada, la inicialización equivale a:

```
Identificador.__init__(persona, "Luis", 29)
````

Por lo tanto:

- `self.nombre` almacena "Luis".
- `self.edad` almacena 29.

`````python
print(persona.nombre)
print(persona.edad)
`````

Cada objetivo tiene sus propios atributos, por ejemplo:

`````python
persona1 = Identificador("Luis", 29)
persona2 = Identificador("Daniel", 27)
`````

`````python
# Ambos objetos pertenecen a la clase Identificador, pero contienen valores diferentes:
print(persona1.nombre)  # Luis
print(persona2.nombre)  # Daniel
`````

Esto significa que una clase define la estructura general, mientras que cada instancia contiene sus propios datos.

#### 3. Métodos: comportamiento del objeto

Los métodos son funciones definidas dentro de una clase que representan las acciones o comportamientos de sus objetos.

- Se definen mediante def, igual que las funciones.
- Los métodos de instancia reciben self como primer parámetro.
- Pueden consultar o modificar los atributos del propio objeto.
- Se ejecutan utilizando la notación objeto.metodo().
- Por ejemplo, agreguemos un método saludar() a nuestra clase:

`````python
class Identificador:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        print(f"Hola, soy {self.nombre} y tengo {self.edad} años.")
`````

`````python
persona1 = Identificador("Luis", 29)
persona2 = Identificador("Daniel", 27)

persona1.saludar()
persona2.saludar()
`````

Observa que `saludar()` es el mismo método para ambos objetos, pero produce resultados diferentes porque utiliza los atributos de cada instancia.

¿Cómo se relacionan `self`, atributos y métodos?

En conjunto:

- Los atributos representan el estado del objeto.
- Los métodos representan su comportamiento.
- self permite acceder al estado y a los métodos de una instancia desde sus propios métodos.

En nuestro ejemplo:


`````python
persona1 = Identificador("Luis", 29)
persona1.saludar()
`````

- Identificador es la clase.
- persona1 es una instancia de esa clase.
- nombre y edad son atributos.
- saludar() es un método.
- self permite que saludar() acceda a los atributos de persona1.

### Otro ejemplo: calcular el área de un rectángulo

Primero una función, después una clase. Las dos alternativas son válidas.

`````python
def calcula_area_pro(largo, ancho):
    area = largo * ancho
    return area

calcula_area_pro(5, 3)
`````

`````python
class Rectangulo:
    def __init__(self, largo, ancho):
        self.largo = largo
        self.ancho = ancho

    def area(self):
        return self.largo * self.ancho

    def perimetro(self):
        return 2 * (self.largo + self.ancho)
`````

`````python
rectangulo = Rectangulo(2, 3)
print("Área:", rectangulo.area())
print("Perímetro:", rectangulo.perimetro())
`````

#### Herencia (extensión opcional)

Una clase hija puede reutilizar el comportamiento de una clase padre. Aquí un cuadrado es un caso especial de rectángulo.



`````python
class Cuadrado(Rectangulo):
    def __init__(self, lado):
        super().__init__(lado, lado)

cuadrado = Cuadrado(4)
print("Área:", cuadrado.area())
print("Perímetro:", cuadrado.perimetro())
`````
